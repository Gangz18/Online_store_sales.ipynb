## PART 5 - Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("sales_data_2000_rows (2).xlsx - Sheet1.csv")

print("Dataset loaded successfully!")
print(df.head())
```

```python
df = df.dropna(subset=["Quantity"])

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df = df.dropna(subset=["Date"])

print("\nMissing values handled!")
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]

print("\nRevenue column created!")
```

```python
city_revenue = df.groupby("City")["Revenue"].sum()

plt.figure(figsize=(10, 5))

city_revenue.plot(kind="bar")

plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Total Revenue (₹)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("01_Revenue_by_City.png")
plt.show()
```

```python
category_revenue = df.groupby("Category")["Revenue"].sum()

plt.figure(figsize=(8, 8))

plt.pie(
    category_revenue,
    labels=category_revenue.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Category Contribution to Total Revenue")

plt.tight_layout()
plt.savefig("02_Category_Contribution.png")
plt.show()
```

```python
df["Month"] = df["Date"].dt.to_period("M")

monthly_sales = df.groupby("Month")["Revenue"].sum()

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("03_Monthly_Sales_Trend.png")
plt.show()
print("\nPART 5 COMPLETED SUCCESSFULLY!")

print("\nCity-wise Revenue:")
print(city_revenue.sort_values(ascending=False))

print("\nCategory-wise Revenue:")
print(category_revenue.sort_values(ascending=False))

print("\nMonthly Sales:")
print(monthly_sales)

print("\nAll three charts saved successfully!")
```
